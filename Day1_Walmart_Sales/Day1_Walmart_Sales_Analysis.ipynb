```python


# 1. We import two libraries: pandas to manage our data table,
#    and a special tool to let us use SQL.
import pandas as pd
from pandasql import sqldf

# 2. This creates our SQL "engine".
run_sql = lambda q: sqldf(q, globals())

# 3. We load the Walmart CSV file into a data table named 'walmart_data'.
walmart_data = pd.read_csv("/kaggle/input/walmart-sales-dataset-of-45stores/walmart-sales-dataset-of-45stores.csv")

print("Setup complete! Here are the first 5 rows of your data:")
print(walmart_data.head())
```

```python
# Query 1: Select all columns and all rows from the table

query1 = """
SELECT *
FROM walmart_data;
"""

# Run the query and display the result
results = run_sql(query1)
print(results)
```

```python
# Query 2: Find all the unique store numbers

query2 = """
SELECT DISTINCT Store
FROM walmart_data;
"""

results2 = run_sql(query2)
print(results2)
```

```python
# Query 3: Select all data only where the store is 20

query3 = """
SELECT *
FROM walmart_data
WHERE Store = 20;
"""

results3 = run_sql(query3)
print(results3)
```

```python
# Query 4: Select all data for Store 20, and sort it by weekly sales (highest first)

query4 = """
SELECT *
FROM walmart_data
WHERE Store = 20
ORDER BY Weekly_Sales DESC;
"""

results4 = run_sql(query4)
print(results4)
```

```python
# Query 5: Calculate company-wide summary statistics

query5 = """
SELECT
    SUM(Weekly_Sales) AS Total_Revenue,
    AVG(Weekly_Sales) AS Average_Weekly_Sales,
    MAX(Weekly_Sales) AS Best_Sales_Week,
    COUNT(*) AS Number_of_Sales_Entries
FROM walmart_data;
"""

results5 = run_sql(query5)
print(results5)
```

```python
# Query 6: Calculate the total sales for each store and rank them

query6 = """
SELECT
    Store,
    SUM(Weekly_Sales) AS Total_Store_Revenue
FROM walmart_data
GROUP BY Store
ORDER BY Total_Store_Revenue DESC;
"""

results6 = run_sql(query6)
print(results6)
```
