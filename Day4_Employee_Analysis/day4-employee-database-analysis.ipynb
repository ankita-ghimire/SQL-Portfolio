```python

import pandas as pd
from pandasql import sqldf
run_sql = lambda q: sqldf(q, globals())

departments_data = {
    'department_id': [1, 2, 3, 4],
    'department_name': ['Human Resources', 'Engineering', 'Marketing', 'Sales']
}
Departments = pd.DataFrame(departments_data)


employees_data = {
    'employee_id': [101, 102, 103, 104, 105],
    'first_name': ['Alice', 'Bob', 'Charlie', 'David', 'Frank'],
    'last_name': ['Smith', 'Johnson', 'Williams', 'Jones', 'Brown'],
    'department_id': [2, 2, 3, 4, 2],
    'manager_id': [105, 105, 104, None, None]
}
Employees = pd.DataFrame(employees_data)


salaries_data = {
    'employee_id': [101, 102, 103, 104, 105],
    'salary': [90000, 95000, 70000, 80000, 150000]
}
Salaries = pd.DataFrame(salaries_data)

print("--- Departments Table ---")
print(Departments)
print("\n--- Employees Table ---")
print(Employees)
print("\n--- Salaries Table ---")
print(Salaries)
print("\nSetup complete! You have three tables: 'Departments', 'Employees', and 'Salaries'.")
```

```python

query1 = """
SELECT
    e.first_name,
    e.last_name,
    d.department_name,
    s.salary
FROM
    Employees AS e
INNER JOIN
    Departments AS d ON e.department_id = d.department_id
INNER JOIN
    Salaries AS s ON e.employee_id = s.employee_id;
"""
results1 = run_sql(query1)
print(results1)
```

```python
# Query 2: Use string functions to format employee names and create an email
query2 = """
SELECT
    e.first_name || ' ' || e.last_name AS full_name,
    LOWER(e.first_name || '.' || e.last_name) || '@company.com' AS company_email,
    d.department_name,
    s.salary
FROM
    Employees AS e
INNER JOIN
    Departments AS d ON e.department_id = d.department_id
INNER JOIN
    Salaries AS s ON e.employee_id = s.employee_id;
"""

results2 = run_sql(query2)
print(results2)
```

```python
query3 = """
SELECT
    e.first_name || ' ' || e.last_name AS employee_name,
    m.first_name || ' ' || m.last_name AS manager_name
FROM
    Employees AS e
LEFT JOIN
    Employees AS m ON e.manager_id = m.employee_id;
"""

results3 = run_sql(query3)
print(results3)
```
