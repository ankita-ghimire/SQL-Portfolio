```python
# 1. We  use pandas and our SQL tool
import pandas as pd
from pandasql import sqldf
run_sql = lambda q: sqldf(q, globals())

# 2. Create the Customers table
customers_data = {
    'customer_id': [101, 102, 103, 104, 105],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'city': ['New York', 'Los Angeles', 'Chicago', 'New York', 'Chicago']
}
Customers = pd.DataFrame(customers_data)

# 3. Create the Orders table
orders_data = {
    'order_id': [1, 2, 3, 4, 5, 6, 7],
    'customer_id': [101, 102, 101, 103, 104, 102, 101],
    'order_amount': [250, 150, 75, 300, 500, 50, 125]
}
Orders = pd.DataFrame(orders_data)

print("--- Customers Table ---")
print(Customers)
print("\n--- Orders Table ---")
print(Orders)
print("\nSetup complete! You now have two tables: 'Customers' and 'Orders'.")
```

```python
# Query 1: Combine the Customers and Orders tables
query1 = """
SELECT
    o.order_id,
    o.order_amount,
    c.name,
    c.city
FROM
    Orders AS o
INNER JOIN
    Customers AS c ON o.customer_id = c.customer_id;
"""

results1 = run_sql(query1)
print(results1)
```

```python
# Query 2: Calculate the total amount spent by each customer
query2 = """
SELECT
    c.name,
    SUM(o.order_amount) AS total_spent
FROM
    Orders AS o
INNER JOIN
    Customers AS c ON o.customer_id = c.customer_id
GROUP BY
    c.name
ORDER BY
    total_spent DESC;
"""

results2 = run_sql(query2)
print(results2)
```

```python
# Query 3: Find customers who have spent more than $400 in total
query3 = """
SELECT
    c.name,
    SUM(o.order_amount) AS total_spent
FROM
    Orders AS o
INNER JOIN
    Customers AS c ON o.customer_id = c.customer_id
GROUP BY
    c.name
HAVING
    SUM(o.order_amount) > 400
ORDER BY
    total_spent DESC;
"""

results3 = run_sql(query3)
print(results3)
```
