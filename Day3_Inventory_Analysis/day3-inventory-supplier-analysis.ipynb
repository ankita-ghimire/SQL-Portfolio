```python


import pandas as pd
from pandasql import sqldf
run_sql = lambda q: sqldf(q, globals())

# Create the Products table
# Note: The "Drone" has no supplier_id
products_data = {
    'product_id': [1, 2, 3, 4],
    'product_name': ['Laptop', 'Mouse', 'Keyboard', 'Drone'],
    'supplier_id': [10, 10, 20, None]
}
Products = pd.DataFrame(products_data)

# Create the Suppliers table
# Note: "Global Parts Inc." does not supply any of our products (intentional)
suppliers_data = {
    'supplier_id': [10, 20, 30],
    'supplier_name': ['TechCorp', 'KeyMasters', 'Global Parts Inc.']
}
Suppliers = pd.DataFrame(suppliers_data)

# Create the Orders table with dates
orders_data = {
    'order_id': [1001, 1002, 1003, 1004, 1005],
    'product_id': [1, 2, 1, 4, 3],
    'order_date': ['2024-01-15', '2024-03-22', '2025-02-10', '2025-04-05', '2024-11-20']
}
Orders = pd.DataFrame(orders_data)


print("--- Products Table ---")
print(Products)
print("\n--- Suppliers Table ---")
print(Suppliers)
print("\n--- Orders Table ---")
print(Orders)
print("\nSetup complete! You have three tables: 'Products', 'Suppliers', and 'Orders'.")
```

```python
# Query 1: Show ALL products and their corresponding supplier, if one exists
query1 = """
SELECT
    p.product_name,
    s.supplier_name
FROM
    Products AS p
LEFT JOIN
    Suppliers AS s ON p.supplier_id = s.supplier_id;
"""

results1 = run_sql(query1)
print(results1)
```

```python
# Query 2: Show ONLY the products that are missing a supplier
query2 = """
SELECT
    p.product_name
FROM
    Products AS p
LEFT JOIN
    Suppliers AS s ON p.supplier_id = s.supplier_id
WHERE
    s.supplier_name IS NULL;
"""

results2 = run_sql(query2)
print(results2)
```

```python
# Query 3 (Corrected): Show ALL suppliers using a LEFT JOIN
query3_corrected = """
SELECT
    p.product_name,
    s.supplier_name
FROM
    Suppliers AS s
LEFT JOIN
    Products AS p ON s.supplier_id = p.supplier_id;
"""

results3_corrected = run_sql(query3_corrected)
print(results3_corrected)
```

```python
# Query 4: Count the number of orders per year
query4 = """
SELECT
    strftime('%Y', order_date) AS order_year,
    COUNT(order_id) AS number_of_orders
FROM
    Orders
GROUP BY
    order_year
ORDER BY
    order_year;
"""

results4 = run_sql(query4)
print(results4)
```
